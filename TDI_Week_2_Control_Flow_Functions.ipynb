## Task 1: Building an Order Approval Function

### Business Scenario

A company processes purchase orders from different suppliers. To maintain financial control, the company uses approval thresholds to determine which orders require management review.

The objective is to create a reusable Python function that automatically determines the appropriate approval level based on the purchase order value.

### Business Rules

| Order Value | Approval Decision |
|---|---|
| Above 100,000 | Needs manager approval |
| Above 50,000 | Needs supervisor approval |
| 50,000 or below | Auto-approved |

### Expected Outcome

Create a function called `approval_message()` that accepts an order value and returns the appropriate approval message.

```python
def approval_message(order_value):

    if order_value > 100000:
        return "Needs manager approval"

    elif order_value > 50000:
        return "Needs supervisor approval"

    else:
        return "Auto-approved"

```

```python
approval_message(72500)
```

## Task 2: Processing Multiple Orders Using a For Loop

### Business Scenario

The objective is to use a Python `for` loop to process a list of purchase orders and apply the `approval_message()` function created in Task 1.

### Expected Outcome

Python should examine each order value and display the appropriate approval decision automatically.

```python
order_values = [12000, 55000, 150000, 49999, 100001, 8000]
```

```python
for order_value in order_values:
    print("Order Value:", order_value)
    print("Approval Decision:", approval_message(order_value))
    print("----------------------------")
```

## Task 3: Supplier Reliability Analysis

### Business Scenario

Supplier performance is important to supply chain efficiency. Suppliers who consistently deliver orders on time help businesses maintain reliable operations.

The objective is to create a Python function that evaluates a supplier's on-time delivery rate and classifies the supplier based on predefined business rules.

### Supplier Classification Rules

- 90% or above: Reliable
- 80% or above but below 90%: Monitor
- Below 80%: At Risk

### Expected Outcome

Create a function called `check_supplier_status()` that accepts a supplier name and on-time delivery rate and returns the appropriate supplier classification.

```python
def check_supplier_status(supplier_name, on_time_rate):

    if on_time_rate >= 0.90:
        return supplier_name + " - Reliable"

    elif on_time_rate >= 0.80:
        return supplier_name + " - Monitor"

    else:
        return supplier_name + " - At Risk"
```

```python
print(check_supplier_status("SupplierA", 0.94))
print(check_supplier_status("SupplierB", 0.86))
print(check_supplier_status("SupplierE", 0.74))
```

## Task 5: Supply Chain Dataset Exploration

### 5.1 Unique Supplier Names

The objective is to identify the different supplier names recorded in the dataset and compare the result with the company's stated number of five suppliers.

This investigation will also help identify possible inconsistencies in supplier naming conventions.

```python
suppliers = set()

for row in rows:
    suppliers.add(row["supplier_name"])

print("Unique supplier names:", suppliers)
print("Number of unique supplier names:", len(suppliers))
```

**Finding: Supplier Name Inconsistency**

The dataset contains 15 unique supplier-name variations, while the company states that it works with five suppliers.

Initial inspection suggests that differences in spacing and letter casing are responsible for the additional variations.

This inconsistency could affect supplier-level reporting and performance analysis if not addressed during a future data-cleaning stage.

No corrections were made during this exploration.

### 5.2 Highest and Lowest Purchase Order Costs

The objective is to identify the highest and lowest purchase order costs recorded in the supply chain dataset.

Because CSV files are initially read as text, the `total_cost` values must be converted into numeric values using Python's `float()` function before performing these comparisons.

```python
total_costs = []

for row in rows:
    cost = float(row["total_cost"])
    total_costs.append(cost)

highest_cost = max(total_costs)
lowest_cost = min(total_costs)

print("Highest order cost:", highest_cost)
print("Lowest order cost:", lowest_cost)
```

**Finding: Purchase Order Cost Investigation**

The highest recorded purchase order cost is 1,720,595.40, while the lowest recorded cost is approximately -207.32.

The negative purchase order value is an unusual observation that may indicate a data-quality issue or require further business investigation.

This finding highlights the importance of validating financial values before using the dataset for business reporting.

No values were modified or removed during this exploration.

### 5.3 Count Delivered Orders

The objective is to determine how many purchase orders in the dataset have a delivery status of `Delivered`.

This will provide an initial understanding of delivery outcomes across the recorded purchase orders.

```python
delivered_orders = 0

for row in rows:
    if row["delivery_status"] == "Delivered":
        delivered_orders += 1

print("Total delivered orders:", delivered_orders)
```

**Finding: Delivery Status Analysis**

The dataset contains 3,000 purchase orders, of which 2,340 have a delivery status of `Delivered`.

This represents 78% of all recorded purchase orders. The remaining 660 orders (22%) have other delivery statuses.

This finding provides an initial overview of delivery outcomes and highlights an opportunity for further investigation into delayed and other non-delivered orders.

```python
delivery_percentage = (delivered_orders / len(rows)) * 100

print("Delivery percentage:", delivery_percentage, "%")
```
