## Task 1: Organising Order Costs Using Lists

### 1.1 Objective

The objective of this task is to use Python lists to organise numerical order costs from the supply chain dataset and perform basic financial analysis.

### 1.2 Data Preparation

The `csv` module was used to read the supply chain dataset. The `total_cost` values were extracted from each order, converted into numerical values using `float()`, and stored in a Python list.

### 1.3 Financial Analysis

The following metrics were calculated:

- Total supply chain spend.
- Average order value.
- Median order value.
- Five smallest order costs.
- Five largest order costs.

### 1.4 Initial Findings

The dataset contains 3,000 orders, with a total recorded order cost of 170,038,100.28.

The average order value is 56,679.37, while the median order value is 30,911.19. The difference suggests that relatively high-cost orders may be influencing the average.

Further examination of the smallest and largest orders will help us understand the distribution of order costs.

```python
import csv

# Create an empty list to store order costs
costs = []

# Open the supply chain dataset
with open("supply_chain_dataset.csv", "r", newline="", encoding="utf-8") as file:

    reader = csv.DictReader(file)

    # Read each order from the dataset
    for row in reader:

        # Convert total_cost from text to a number
        cost = float(row["total_cost"])

        # Add the cost to our list
        costs.append(cost)

# Display the results
print("Total number of orders:", len(costs))
print("First five order costs:", costs[:5])
```

```python
# Calculate the total cost of all orders
total_spend = sum(costs)

# Calculate the average order value
average_order_value = total_spend / len(costs)

# Display the results
print("Total supply chain spend:", round(total_spend, 2))
print("Average order value:", round(average_order_value, 2))
```

```python
# Sort order costs from smallest to largest
sorted_costs = sorted(costs)

# Calculate the median
middle = len(sorted_costs) // 2

median_order_value = (
    sorted_costs[middle - 1] + sorted_costs[middle]
) / 2

print("Median order value:", round(median_order_value, 2))
```

```python
# Get the five smallest order costs
smallest_orders = sorted_costs[:5]

# Get the five largest order costs
largest_orders = sorted_costs[-5:]

print("Five smallest order costs:")
print(smallest_orders)

print("\nFive largest order costs:")
print(largest_orders)
```

### 1.5 Smallest and Largest Order Costs

The five smallest and largest order costs were identified using Python list slicing after sorting the order-cost list.

**Five smallest order costs:**
-207.32, 335.27, 340.35, 340.52 and 352.60.

**Five largest order costs:**
765,820.16, 807,800.07, 991,898.31, 1,000,681.24 and 1,720,595.40.

### 1.6 Key Observations

1. The dataset contains a negative order cost of -207.32. This value requires further investigation to determine whether it represents a legitimate financial adjustment or a data-quality issue.
2. The largest order cost is 1,720,595.40, which is substantially higher than the median order value of 30,911.19.
3. The difference between the average and median, alongside the high-value orders, suggests that the order-cost distribution may be influenced by unusually large transactions.

These observations will be useful when assessing supplier spending and potential data-quality issues in subsequent tasks.

## Task 2: Mapping Product Categories to Suppliers

### 2.1 Objective

The objective of this task is to use Python dictionaries and sets to map product categories to their associated suppliers.

This structure will help identify supplier relationships, determine how many suppliers serve each category, and support the assessment of single-source supply risks.

### 2.2 Data Standardisation

Helper functions will be used to standardise supplier names and remove leading or trailing whitespace from product categories for analysis.

The original dataset will remain unchanged.

```python
# Standardise supplier names
def clean_supplier(name):
    compact = name.strip().lower().replace(" ", "")
    return "Supplier" + compact[-1].upper()


# Standardise product category names
def clean_category(category):
    return category.strip()
```

```python
# Create an empty dictionary
category_suppliers = {}

# Read the dataset
with open("supply_chain_dataset.csv", "r", newline="", encoding="utf-8") as file:

    reader = csv.DictReader(file)

    for row in reader:

        # Standardise supplier and category names
        supplier = clean_supplier(row["supplier_name"])
        category = clean_category(row["product_category"])

        # Create an empty set for a new category
        if category not in category_suppliers:
            category_suppliers[category] = set()

        # Add supplier to the category's set
        category_suppliers[category].add(supplier)

# Display the dictionary
print(category_suppliers)
```

```python
# Retrieve suppliers for Pumps
pump_suppliers = category_suppliers["Pumps"]

print("Suppliers providing Pumps:", pump_suppliers)
print("Number of suppliers:", len(pump_suppliers))
```

```python
# Create an empty set for all suppliers
all_suppliers = set()

# Collect suppliers from every product category
for suppliers in category_suppliers.values():
    all_suppliers.update(suppliers)

# Display the unique suppliers
print("Unique suppliers:", all_suppliers)
print("Total unique suppliers:", len(all_suppliers))
```

## Task 3: Exploring Unique Values Using Sets

### 3.1 Objective

The objective of this task is to use Python sets to identify unique suppliers, product categories and delivery statuses in the supply chain dataset.

The analysis will compare raw unique values with standardised supplier and category values to identify inconsistencies in the dataset.

Delivery statuses will also be examined for possible spelling or formatting variations.

```python
# Create empty sets
raw_suppliers = set()
raw_categories = set()
delivery_statuses = set()

# Read the dataset
with open("supply_chain_dataset.csv", "r", newline="", encoding="utf-8") as file:

    reader = csv.DictReader(file)

    for row in reader:

        # Collect raw unique values
        raw_suppliers.add(row["supplier_name"])
        raw_categories.add(row["product_category"])
        delivery_statuses.add(row["delivery_status"])

# Display unique counts
print("Raw unique suppliers:", len(raw_suppliers))
print("Raw unique categories:", len(raw_categories))
print("Unique delivery statuses:", len(delivery_statuses))
```

```python
# Create sets for standardised values
clean_suppliers = set()
clean_categories = set()

# Standardise the raw values
for supplier in raw_suppliers:
    clean_suppliers.add(clean_supplier(supplier))

for category in raw_categories:
    clean_categories.add(clean_category(category))

# Compare the unique counts
print("Supplier comparison")
print("Before standardisation:", len(raw_suppliers))
print("After standardisation:", len(clean_suppliers))

print("\nCategory comparison")
print("Before standardisation:", len(raw_categories))
print("After standardisation:", len(clean_categories))
```

```python
# Display all unique delivery statuses
print("Unique delivery statuses:")

for status in sorted(delivery_statuses):
    print(status)
```

**Observations:**

## Task 4: Identifying Single-Source Supply Risks

### 4.1 Objective

The objective of this task is to identify product categories that depend on only one recorded supplier.

Categories with a single supplier may be more vulnerable to supply interruptions because the dataset does not show an alternative sourcing option.

The analysis will use dictionary iteration and conditional statements to identify these categories and calculate their proportion of the total product categories.

```python
# Initialise a counter
single_source_count = 0

# Examine each product category
for category, suppliers in category_suppliers.items():

    # Check whether the category has exactly one supplier
    if len(suppliers) == 1:

        single_source_count += 1

        # Display the category and its supplier
        print(category, "->", suppliers)

# Calculate total number of categories
total_categories = len(category_suppliers)

# Display the summary
print("\nSingle-source categories:",
      single_source_count, "of", total_categories)

print("Single-source percentage:",
      round((single_source_count / total_categories) * 100, 2), "%")
```

The analysis identified four product categories with exactly one recorded supplier.

| Product Category | Sole Supplier |
|---|---|
| Fasteners | SupplierE |
| Control Systems | SupplierA |
| Connectors | SupplierE |
| Hydraulics | SupplierD |

**Summary:**

- Total product categories: 8
- Single-source categories: 4
- Single-source percentage: 50%

### 4.3 Business Risk Interpretation

The analysis shows that 50% of the product categories depend on a single recorded supplier.

SupplierE is the sole supplier for both Fasteners and Connectors. This creates a potential concentration of supply dependency because disruption involving SupplierE could affect two categories simultaneously.

SupplierA and SupplierD are also sole suppliers for Control Systems and Hydraulics, respectively.

These dependencies may expose the company to supply interruptions if the affected suppliers cannot fulfil orders. Management could investigate alternative suppliers, supplier capacity and contingency sourcing arrangements to improve supply continuity.

The findings represent potential risks based on the available supplier-category relationships, rather than evidence of actual supplier failures.

## Task 5: Building the Supplier Network

### 5.1 Objective

The objective of this task is to build a structured supplier network using nested dictionaries and sets.

Each supplier will have three attributes:

- Product categories supplied.
- Total number of orders.
- Total recorded order spend.

This structure will support supplier-level analysis and help identify suppliers with broad category coverage and higher financial exposure.

```python
# Create an empty supplier network
network = {}

# Initialise each supplier
for supplier in all_suppliers:

    network[supplier] = {
        "categories": set(),
        "order_count": 0,
        "total_spend": 0.0
    }

# Display the initial network
print(network)
```

```python
# Read the dataset and populate the supplier network
with open("supply_chain_dataset.csv", "r", newline="", encoding="utf-8") as file:

    reader = csv.DictReader(file)

    for row in reader:

        # Standardise supplier and category names
        supplier = clean_supplier(row["supplier_name"])
        category = clean_category(row["product_category"])

        # Convert order cost into a number
        cost = float(row["total_cost"])

        # Add the category to the supplier's category set
        network[supplier]["categories"].add(category)

        # Increase the supplier's order count
        network[supplier]["order_count"] += 1

        # Add the order cost to the supplier's total spend
        network[supplier]["total_spend"] += cost

print("Supplier network populated successfully.")
```

```python
# Retrieve SupplierA's complete information
supplier_a = network["SupplierA"]

print("SUPPLIER A NETWORK REPORT")
print("-------------------------")

print("Categories:", supplier_a["categories"])
print("Number of orders:", supplier_a["order_count"])
print("Total spend:", round(supplier_a["total_spend"], 2))
```

```python
print("COMPLETE SUPPLIER NETWORK REPORT")
print("=" * 65)

for supplier, details in network.items():
    print(f"\nSupplier: {supplier}")
    print("Categories:", details["categories"])
    print("Number of categories:", len(details["categories"]))
    print("Number of orders:", details["order_count"])
    print("Total spend:", round(details["total_spend"], 2))
    print("-" * 65)
```

```python
print("SUPPLIER NETWORK VERIFICATION")
print("=" * 75)

total_network_orders = 0

for supplier, details in sorted(network.items()):
    total_network_orders += details["order_count"]

    print(
        f"{supplier:<12} | "
        f"Orders: {details['order_count']:<5} | "
        f"Categories: {sorted(details['categories'])} | "
        f"Category count: {len(details['categories'])} | "
        f"Spend: {details['total_spend']:,.2f}"
    )

print("=" * 75)
print("Total orders across suppliers:", total_network_orders)
print("Total orders in dataset:", len(costs))
```

```python
status_costs = {}

with open("supply_chain_dataset.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        status = row["delivery_status"]
        cost = float(row["total_cost"])

        if status not in status_costs:
            status_costs[status] = []

        status_costs[status].append(cost)

print("AVERAGE ORDER VALUE BY DELIVERY STATUS")
print("=" * 55)

for status, values in status_costs.items():
    average_cost = sum(values) / len(values)

    print(
        f"{status:<15} | "
        f"Orders: {len(values):<5} | "
        f"Average order value: {average_cost:,.2f}"
    )
```
